```python
import pandas as pd
import numpy as np

np.random.seed(42)
sessions = 250  # each row = one 15-minute test session

# ── VFD FREQUENCY ─────────────────────────────────────────
vfd_freq = np.concatenate([
    np.random.choice([50, 40, 35], size=210, p=[0.6, 0.3, 0.1]),
    np.random.choice([50, 40, 35], size=40,  p=[0.6, 0.3, 0.1])
])

# ── EXPECTED SPEED from VFD ───────────────────────────────
# 50Hz → 0.8 m/s, 40Hz → 0.64 m/s, 35Hz → 0.56 m/s
expected_speed = (vfd_freq / 50) * 0.8

# ── NORMAL SESSIONS (1-210) ───────────────────────────────
noise_normal       = np.random.normal(0, 0.02, 210)
speed_normal       = expected_speed[:210] + noise_normal
temperature_normal = np.random.normal(55, 4, 210)
current_normal     = np.random.normal(2.1, 0.2, 210)
items_normal       = np.random.randint(25, 50, 210)

# ── ANOMALY SESSIONS (211-250) ────────────────────────────
# gradual degradation — belt slipping, motor heating, current rising
# speed drops gradually but never reaches 0 (VFD protects motor)
noise_wear       = np.random.normal(0, 0.02, 40)
speed_wear       = expected_speed[210:] + noise_wear - np.linspace(0, 0.35, 40)
speed_wear       = np.clip(speed_wear, 0.15, 0.8)  # VFD min protection

temperature_wear = np.random.normal(55, 4, 40) + np.linspace(0, 25, 40)
temperature_wear = np.clip(temperature_wear, 55, 82)  # realistic motor limit

# current rises but never exceeds 3.3A (VFD overcurrent trips before 3.5A)
current_wear     = np.random.normal(2.1, 0.2, 40) + np.linspace(0, 1.1, 40)
current_wear     = np.clip(current_wear, 2.1, 3.3)

items_wear       = np.random.randint(25, 50, 40) - np.linspace(0, 20, 40).astype(int)
items_wear       = np.clip(items_wear, 5, 50)

# ── COMBINE ───────────────────────────────────────────────
belt_speed  = np.concatenate([speed_normal,       speed_wear])
temperature = np.concatenate([temperature_normal, temperature_wear])
current     = np.concatenate([current_normal,     current_wear])
items       = np.concatenate([items_normal,       items_wear])

# ── SPEED DEVIATION ───────────────────────────────────────
speed_deviation = (belt_speed - expected_speed).round(3)

# ── MAINTENANCE LOGS ──────────────────────────────────────
# technician comes every 15 sessions normally
maintenance_sessions = [15, 30, 45, 60, 75, 90, 105, 120,
                        135, 150, 165, 180, 195, 210, 228]

maintenance_done = np.zeros(sessions, dtype=int)
for s in maintenance_sessions:
    maintenance_done[s - 1] = 1

days_since_maint = np.zeros(sessions)
counter = 0
for i in range(sessions):
    if maintenance_done[i] == 1:
        counter = 0
    else:
        counter += 1
    days_since_maint[i] = counter

interventions_30d = np.zeros(sessions)
for i in range(sessions):
    start = max(0, i - 29)
    interventions_30d[i] = maintenance_done[start:i+1].sum()

# ── RUL ───────────────────────────────────────────────────
rul = np.zeros(sessions)
cycle = 0
for i in range(sessions):
    if maintenance_done[i] == 1:
        cycle = 0
    if i < 210:
        rul[i] = max(0, 90 - cycle * 2)
    else:
        rul[i] = max(0, 90 - cycle * 4)
    cycle += 1

# ── ALERTS ────────────────────────────────────────────────
alerts = np.zeros(sessions, dtype=int)
for i in range(sessions):
    if i < 210:
        alerts[i] = np.random.randint(0, 3)
    else:
        alerts[i] = np.random.randint(3, 10)
    if i > 0 and maintenance_done[i-1] == 1:
        alerts[i] = max(0, alerts[i] - 2)

# ── BUILD DATAFRAME ───────────────────────────────────────

session_num = range(1, sessions + 1)
df = pd.DataFrame({
    'session':                session_num,
    'belt_speed':             belt_speed.round(3),
    'vfd_frequency':          vfd_freq,
    'expected_speed':         expected_speed.round(3),
    'speed_deviation':        speed_deviation,
    'temperature':            temperature.round(1),
    'current':                current.round(2),
    'items_count':            items.astype(int),
    'maintenance_done_today': maintenance_done,
    'days_since_maintenance': days_since_maint.astype(int),
    'interventions_30d':      interventions_30d.astype(int),
    'rul':                    rul.round(1),
    'alerts_today':           alerts
})

df.to_csv('../data/sensor_data.csv', index=False)
print("Done! Shape:", df.shape)
print("\n--- Normal sessions sample ---")
print(df[['session','belt_speed','speed_deviation',
          'temperature','current','items_count']].head(5))
print("\n--- Anomaly sessions last 5 ---")
print(df[['session','belt_speed','speed_deviation',
          'temperature','current','items_count']].tail(5))
```
